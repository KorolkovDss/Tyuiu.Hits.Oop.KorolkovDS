### Вариант задания 16


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс PaymentMethod в C#, который будет представлять
различные способы оплаты. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

<h2 style="color:DodgerBlue">Требования к базовому классу PaymentMethod:</h2>

• Атрибуты: ID способа оплаты (PaymentMethodId), Название способа оплаты
(MethodName), Минимальная сумма (MinAmount).
• Методы:
o ProcessPayment(decimal amount): метод для обработки платежа
указанной суммы.
o CheckMinimumAmount(decimal amount): метод для проверки
минимальной суммы платежа.
o GetPaymentDetails(): метод для получения деталей способа оплаты.

<h2 style="color:DodgerBlue">Требования к производным классам:</h2>

1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные
атрибуты, такие как URL платежной системы (PaymentUrl).
Метод ProcessPayment() должен быть переопределен для включения URL
платежной системы в процесс оплаты.
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные
атрибуты, такие как Банковские данные (BankData).
Метод CheckMinimumAmount() должен быть переопределен для проверки
минимальной суммы платежа с учетом банковских комиссий.
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Место выдачи наличных
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен
для отображения места выдачи наличных.



<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class PaymentMethod
{
    public int PaymentMethodId;
    public string MethodName;
    public decimal MinAmount;

    
    public virtual void ProcessPayment(decimal amount)
    {
        Console.WriteLine($"Платеж на сумму {amount} руб. обработан");
    }

    
    public virtual void CheckMinimumAmount(decimal amount)
    {
        if (amount >= MinAmount)
            Console.WriteLine($"Сумма {amount} руб. подходит (минимум {MinAmount} руб.)");
        else
            Console.WriteLine($"Сумма {amount} руб. слишком мала (нужно минимум {MinAmount} руб.)");
    }

    
    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"ID: {PaymentMethodId}, Название: {MethodName}, Мин. сумма: {MinAmount} руб.");
    }
}


public class OnlinePayment : PaymentMethod
{
    public string PaymentUrl;

    
    public override void ProcessPayment(decimal amount)
    {
        base.ProcessPayment(amount);
        Console.WriteLine($"Ссылка на оплату: {PaymentUrl}");
    }
}


public class BankTransfer : PaymentMethod
{
    public string BankData;

    
    public override void CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount * 0.015m; // 1.5% комиссия
        decimal finalAmount = amount - commission;
        
        Console.WriteLine($"Банк: {BankData}");
        Console.WriteLine($"Комиссия: {commission:F2} руб.");
        Console.WriteLine($"К зачислению: {finalAmount:F2} руб.");
        
        if (finalAmount >= MinAmount)
            Console.WriteLine($"Сумма подходит (с учетом комиссии)");
        else
            Console.WriteLine($"Сумма слишком мала (с учетом комиссии)");
    }
}


public class CashPayment : PaymentMethod
{
    public string CashPickupPoint;

    
    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"Место выдачи: {CashPickupPoint}");
    }
}


OnlinePayment online = new OnlinePayment();
online.PaymentMethodId = 1;
online.MethodName = "Оплата картой";
online.MinAmount = 100;
online.PaymentUrl = "https://pay.example.com";

online.GetPaymentDetails();
online.CheckMinimumAmount(150);
online.ProcessPayment(150);
Console.WriteLine();


BankTransfer bank = new BankTransfer();
bank.PaymentMethodId = 2;
bank.MethodName = "Перевод по реквизитам";
bank.MinAmount = 500;
bank.BankData = "Сбербанк, счет 1234567890";

bank.GetPaymentDetails();
bank.CheckMinimumAmount(600);
bank.ProcessPayment(600);
Console.WriteLine();


CashPayment cash = new CashPayment();
cash.PaymentMethodId = 3;
cash.MethodName = "Наличные";
cash.MinAmount = 50;
cash.CashPickupPoint = "ул. Тверская, д. 1";

cash.GetPaymentDetails();
cash.CheckMinimumAmount(75);
cash.ProcessPayment(75);
```
